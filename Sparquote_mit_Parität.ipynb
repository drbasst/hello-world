# **Sparquote mit Parität der Konsumkraft**

# **Quellcode**

```python
import numpy as np
import itertools as it
import matplotlib.pyplot as plt

def calc_SavingsRate(r_list, i_list, t_list, l_list, p_list, s_list):
    rates = []

    for c in it.product(r_list, i_list, t_list, l_list, p_list, s_list):
        r, i, t, l, p, s = c

        d1 = p/l/(1 + i/100)**t
        d2 = 1 - s/100*(1 - r/100*t/((1 + r/100)**t - 1)/(1 + r/100))
        d3 = ((1 + r/100)**t - 1)/(r/100)*(1 + r/100)
        rates.append(100/(d1*d2*d3 + 1))

    return rates

def calc_Stats(rates):
    quants = np.quantile(rates, [0.10, 0.25, 0.50, 0.75, 0.90])
    bins = np.arange(np.round(quants[2],-1)-30, np.round(quants[2],-1)+31, 1)
    hist = np.histogram(rates, bins, density=True)[0]

    return [quants, bins, hist]

def plot_Hist(rates, stats):
    size = 18
    string = 'Quantile:\n$s_{90}$ = %.2f %%\n$s_{75}$ = %.2f %%\n$s_{50}$ = %.2f %%\n$s_{25}$ = %.2f %%\n$s_{10}$ = %.2f %%'%(tuple(stats[0][::-1]))

    fig = plt.figure(figsize=(16,9), tight_layout=True)
    ax = fig.add_subplot(111)
    ax.bar(stats[1][:-1], 100*stats[2], width=0.90, align='edge', color='tab:blue', label='Rel. Häufigkeit')
    ax.text(stats[1][0]+1, 8.5, s=string, fontsize=size, va='center', ha='left', bbox=dict(boxstyle='round', facecolor='white', alpha=0.75))
    ax.grid()
    ax.legend(loc=0, fontsize=size, title='Sparquote mit Parität:', title_fontsize=size)
    ax.set_xlabel(r'Sparquote / [%]', fontsize=size)
    ax.set_ylabel(r'Rel. Häufigkeit / [%]', fontsize=size)
    ax.set_xlim((stats[1][0], stats[1][-1]))
    ax.set_ylim((0, 10))
    ax.set_xticks(stats[1][::10])
    ax.set_yticks([0, 2, 4, 6, 8, 10])
    ax.tick_params(axis='both', labelsize=size)
    plt.show()
```

# **Anwendung**

```python
# Mittlere Rendite in % p.a.
r_list = np.linspace(7.0, 11.0, 11) # Alternativ: r_list = [9.0]

# Mittlere Inflation in % p.a.
i_list = np.linspace(2.0, 3.0, 11) # Alternativ: i_list = [2.5]

# Laenge der Ansparphase in Jahren
t_list = np.linspace(28, 32, 11) # Alternativ: t_list = [30]

# Rentenluecke in %
l_list = np.linspace(30, 50, 11) # Alternativ: l_list = [40]

# Entnahmerate in %
p_list = np.linspace(2.0, 5.0, 11) # Alternativ: p_list = [3.5]

# Effektiver Steuersatz in %
s_list = np.linspace(20, 30, 11) # Alternativ: s_list = [25]

# Berechne Sparquote mit Paritaet
rates = calc_SavingsRate(r_list, i_list, t_list, l_list, p_list, s_list)
stats = calc_Stats(rates)

# Plotting
plot_Hist(rates, stats)
```
